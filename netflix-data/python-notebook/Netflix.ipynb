<p>Q2) In which year highest no. of the TV Shows & Movies were released ?</p>

```python
df.dtypes # Release_Date has object datatype
```

```python
# Creating new column to store the Release_Date in date time dtype
df['Date_N'] = pd.to_datetime(df['Release_Date'].str.strip()) # .str.strip() to remove extra spaces from date
```

```python
df.head()
```

```python
df.dtypes
```

```python
# Ploting released movies by year using bar chart
df['Date_N'].dt.year.value_counts()
```

```python
df['Date_N'].dt.year.value_counts().plot(kind = 'bar')
plt.xlabel("Years")
plt.ylabel("No. of Movies Released")
plt.tight_layout()
```
